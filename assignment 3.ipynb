```python
def total_cart_amount(prices):
    total = 0

    for price in prices:
        total += float(price)

    return float(total)

prices = ['199.99', '49', '350.75', '499.50']

result = total_cart_amount(prices)

print("Total Cart Amount: ₹", result)
```
